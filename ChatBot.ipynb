```python
SYMPTOMS = {
    "stress": ["stressed", "stress", "overwhelmed"],
    "anxiety": ["anxious", "nervous", "worried", "panic"],
    "depression": ["sad", "hopeless", "depressed"],
    "sleep": ["insomnia", "can't sleep", "sleep problems"],
    "social": ["alone", "isolated", "lonely"]
}
```

```python
def extract_features(text):

    text = text.lower()

    features = {
        "stress": 0,
        "anxiety": 0,
        "depression": 0,
        "sleep": 0,
        "social": 0
    }

    for category, words in SYMPTOMS.items():

        for word in words:

            if word in text:
                features[category] = 1
                break

    return list(features.values())
```

```python
text = """
I have been feeling stressed lately and
I can't sleep well.
"""

print(extract_features(text))
```

```python
import pandas as pd

data = pd.DataFrame({
    "stress":[1,0,0],
    "anxiety":[1,0,0],
    "depression":[0,1,0],
    "sleep":[1,1,0],
    "social":[0,1,0],
    "result":["Anxiety","Depression","Healthy"]
})
X = data[[
    "stress",
    "anxiety",
    "depression",
    "sleep",
    "social"
]]

y = data["result"]

from sklearn.svm import SVC

model = SVC(probability=True)

model.fit(X, y)
```

```python
message = """
I am stressed and I cannot sleep because
of my exams.
"""
features = extract_features(message)

print(features)
```

```python
prediction = model.predict([features])[0]

print(prediction)
```

```python
import random

responses = {
    "Anxiety":[
        "You appear to be experiencing anxiety-related symptoms.",
        "Your message suggests stress and anxiety.",
        "There are indicators of anxiety in what you've described."
    ],

    "Depression":[
        "Your message suggests symptoms related to depression.",
        "There are signs of depressive feelings.",
        "You may be experiencing depression-related symptoms."
    ],

    "Healthy":[
        "No major mental health concerns were detected.",
        "Your responses appear generally healthy.",
        "You seem emotionally stable from the information provided."
    ]
}

reply = random.choice(responses[prediction])

print(reply)
```
